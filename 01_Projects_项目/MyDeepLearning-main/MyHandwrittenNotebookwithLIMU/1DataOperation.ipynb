我们也可以把多个张量 连结（concatenate） 在一起

```python
#torch.arange(12)：生成 0 到 11
#dtype=torch.float32：把数据类型设成浮点数
#.reshape((3, 4))：改成 3 行 4 列
x = torch.arange(12, dtype=torch.float32).reshape((3,4))

#这是直接手工创建一个 3 x 4 的张量
y = torch.tensor([[1,2,3,4],[2,2,3,4],[3,2,3,11]])

#torch.cat 的意思是：把多个张量拼接起来。
#cat 是 concatenate 的缩写。
#dim=0 表示：按第 0 维拼，也就是“上下拼，增加行数”。
Z = torch.cat((x,y),dim = 0)

#dim=1 表示：按第 1 维拼，也就是“左右拼，增加列数”。
W = torch.cat((x,y),dim=1)
Z,W
```

执行原地操作

```python
Z = torch.zeros_like(x)
print('id(Z):', id(Z))
Z[:] = X + x
print('id(Z):', id(Z))
```

如果在后续计算中没有重复使用 X，我们也可以使用 X[:] = X + Y 或 X += Y 来减少操作的内存开销
